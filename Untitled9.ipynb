```python
from typing import Optional, Any


class NoBinario:
    """Representa um nó em uma estrutura de árvore binária dinâmica."""

    def __init__(
        self,
        valor: Any,
        esquerda: Optional["NoBinario"] = None,
        direita: Optional["NoBinario"] = None,
    ) -> None:
        self.valor: Any = valor
        self.esquerda: Optional[NoBinario] = esquerda
        self.direita: Optional[NoBinario] = direita

    def __repr__(self) -> str:
        return f"NoBinario(valor={self.valor!r})"


def inicializar_arvore_streaming() -> NoBinario:
    # Nível 1: Raiz
    raiz = NoBinario("Streaming")

    # Nível 2: Ramos principais
    raiz.esquerda = NoBinario("Filmes")
    raiz.direita = NoBinario("Séries")

    # Nível 3: Subcategorias de Filmes
    raiz.esquerda.esquerda = NoBinario("Ação")
    raiz.esquerda.direita = NoBinario("Comédia")

    return raiz


if __name__ == "__main__":
    arvore = inicializar_arvore_streaming()

    # Navegação por referência direta com verificação segura de nulidade (Mapeamento de caminhos)
    categoria_acao = (
        arvore.esquerda.esquerda.valor
        if arvore.esquerda and arvore.esquerda.esquerda
        else None
    )
    categoria_comedia = (
        arvore.esquerda.direita.valor
        if arvore.esquerda and arvore.esquerda.direita
        else None
    )

    print(f"Categoria de Filme 1: {categoria_acao}")
    print(f"Categoria de Filme 2: {categoria_comedia}")
```

```python
from typing import Optional, Any


class NoBinario:
    """Estrutura profissional de nó de árvore binária aplicável a Árvores de Decisão."""

    def __init__(
        self,
        valor: Any,
        esquerda: Optional["NoBinario"] = None,
        direita: Optional["NoBinario"] = None,
    ) -> None:
        self.valor: Any = valor
        self.esquerda: Optional[NoBinario] = esquerda
        self.direita: Optional[NoBinario] = direita

    @property
    def eh_folha(self) -> bool:
        """Determina se o nó atual é uma folha (terminal/decisão)."""
        return self.esquerda is None and self.direita is None

    def __repr__(self) -> str:
        return f"NoBinario(valor={self.valor!r}, tipo={'Folha' if self.eh_folha else 'Decisão'})"


def construir_arvore_decisao() -> NoBinario:
    """Montagem determinística da árvore de decisão de recomendação."""
    # Perguntas / Nós Internos
    raiz = NoBinario("Quer um filme emocionante?")
    raiz.esquerda = NoBinario("Gosta de ação?")
    raiz.direita = NoBinario("Prefere rir?")

    # Resultados de Decisão (Folhas)
    # Caminho Esquerda (SIM) / Direita (NÃO)
    raiz.esquerda.esquerda = NoBinario("Missão Impossível")
    raiz.esquerda.direita = NoBinario("Interestelar")
    raiz.direita.esquerda = NoBinario("As Branquelas")
    raiz.direita.direita = NoBinario("O Poderoso Chefão")

    return raiz


if __name__ == "__main__":
    arvore_decisao = construir_arvore_decisao()
    print("Estrutura carregada com sucesso na memória do interpretador!")
    print(f"Nó Raiz: {arvore_decisao}")
```

```python
from typing import Optional, Any


class NoBinario:
    """Nó de Árvore Binária otimizado para motores de recomendação."""

    def __init__(
        self,
        valor: Any,
        esquerda: Optional["NoBinario"] = None,
        direita: Optional["NoBinario"] = None,
    ) -> None:
        self.valor: Any = valor
        self.esquerda: Optional[NoBinario] = esquerda
        self.direita: Optional[NoBinario] = direita

    @property
    def eh_folha(self) -> bool:
        return self.esquerda is None and self.direita is None


def construir_arvore() -> NoBinario:
    raiz = NoBinario("Quer um filme emocionante?")
    raiz.esquerda = NoBinario("Gosta de ação?")
    raiz.direita = NoBinario("Prefere rir?")

    raiz.esquerda.esquerda = NoBinario("Missão Impossível")
    raiz.esquerda.direita = NoBinario("Interestelar")
    raiz.direita.esquerda = NoBinario("As Branquelas")
    raiz.direita.direita = NoBinario("O Poderoso Chefão")

    return raiz


def executar_recomendador(arvore: NoBinario) -> None:
    """Executa o fluxo interativo de navegação estruturada (Loop Iterativo)."""
    atual: NoBinario = arvore

    # Navega enquanto houver decisões intermediárias a tomar
    while not atual.eh_folha:
        resposta = input(f"{atual.valor} (SIM/NÃO): ").strip().upper()

        if resposta == "SIM":
            if atual.esquerda:
                atual = atual.esquerda
        elif resposta in ("NÃO", "NAO"):
            if atual.direita:
                atual = atual.direita
        else:
            print("\n[Aviso] Entrada inválida! Responda apenas com SIM ou NÃO.\n")

    print(f"\n{'='*40}")
    print(f"=> RECOMENDAÇÃO FINAL: {atual.valor}")
    print(f"{'='*40}")


if __name__ == "__main__":
    arvore_decisao = construir_arvore()
    executar_recomendador(arvore_decisao)
```
